# 13 Model 1: Bag of Words
The Bag of Words (BoW) model is a simple text representation technique that converts text into numerical vectors based on word counts. It ignores grammar, word order, and context, focusing only on how frequently words appear.

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
data = [' Most shark attacks occur about 10 feet from the beach since that is where the people are',
        'the efficiency with which he paired the socks in the drawer was quite admirable',
        'carol drank the blood as if she were a vampire',
        'giving directions that the mountains are to the west only works when you can see them',
        'the sign said there was road work ahead so he decided to speed up',
        'the gruff old man sat in the back of the bait shop grumbling to himself as he scooped out a handful of worms']
```

```python
countvec = CountVectorizer()
countvec_fit = countvec.fit_transform(data)
print(countvec_fit.toarray())
```

What fit_transform do?

1. The "Fit" Step (Learning)
The fit part looks at your text data and learns the vocabulary.

It identifies all the unique words across all your documents.

It filters out punctuation (by default) and assigns a specific index number to each unique word.

Analogy: Imagine creating a blank dictionary where every unique word in your text gets its own entry page.

2. The "Transform" Step (Applying)
The transform part takes that newly learned vocabulary and converts your text data into a numerical matrix (often called a Document-Term Matrix).

For every row of text, it counts how many times each word from the vocabulary appears.

It outputs a sparse matrix of numbers, which machine learning algorithms can actually understand.

Analogy: Going through your text and filling in the pages of that dictionary with the actual word counts for each document.

```python
bag_of_words = pd.DataFrame(countvec_fit.toarray(), columns = countvec.get_feature_names_out())
print(bag_of_words)
```
